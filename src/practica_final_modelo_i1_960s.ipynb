## Modelo

```python
from ultralytics import YOLO
from ultralytics import settings
settings.update({"tensorboard":True})

# =====================================================
# CONFIGURACIÓN
# =====================================================

DATASET_YAML = "C:\\dev\\repos\\vision\\practica_final\\UDIT\\data\\train\\images_split\\data.yaml"

MODEL_NAME = "yolov8s.pt"

EPOCHS = 100

IMAGE_SIZE = 960

BATCH_SIZE = 8

PROJECT_NAME = "traffic_detection"

RUN_NAME = "yolov8s_traffic960"

# =====================================================
# CARGAR MODELO
# =====================================================

model = YOLO(MODEL_NAME)

# =====================================================
# ENTRENAMIENTO
# =====================================================

results = model.train(
    data=DATASET_YAML,
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,

    # Optimizador
    optimizer="AdamW",

    # Hardware
    device=0,          # GPU 0
    workers=8,

    # Guardado
    project=PROJECT_NAME,
    name=RUN_NAME,

    # Validación
    val=True,

    # Data augmentation
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,

    degrees=5,
    translate=0.1,
    scale=0.5,
    fliplr=0.5,

    mosaic=1.0,
    mixup=0.1,

    # Early stopping
    patience=10,

    # Logs
    verbose=True
)

# =====================================================
# EVALUACIÓN FINAL
# =====================================================

metrics = model.val()

print("\n========== MÉTRICAS ==========")

print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)

# =====================================================
# INFERENCIA DE PRUEBA
# =====================================================

results = model.predict(
    source="data/train/images_split/images/val",
    conf=0.25,
    save=True
)

print("\nPredicciones guardadas.")
```
